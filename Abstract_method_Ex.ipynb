```python
from abc import ABC, abstractmethod

class bank(ABC):
    def __init__(self,user_id,password,
                 acct_no,acct_holdername,
                 age,location,gender):
        self.user_id = user_id
        self.password = password
        self.acct_no = acct_no
        self.acct_holdername = acct_holdername
        self.age = age
        self.location = location
        self.gender = gender

    @abstractmethod
    def credit_bal(self,Amount):
        pass

    @abstractmethod
    def debit_bal(self,Amount):
        pass

    @abstractmethod
    def transfer_bal(self,Amount,target_acct):
        pass

    @abstractmethod
    def check_bal(self):
        pass

        
```

```python
class saving_acct(bank):
    def __init__(self,user_id,password,
                     acct_no,acct_holdername,
                     age,location,gender,balance,interest_rate = 8.0):
            super().__init__(user_id,password,
                     acct_no,acct_holdername,
                     age,location,gender)
    
            self.interest_rate = interest_rate
            self.balance = balance
    
    def credit_bal(self,Amount):
        if Amount > 0:
    
            self.balance += Amount
            print(f'{Amount} has been credited to your account, acc:{self.acct_no}.' ,f'Your current balance is {self.balance}.')
        else :
            print(f'Invalid input')
         
    def debit_bal(self,Amount):
        if self.balance > Amount :
    
            self.balance-=Amount
            print(f'{Amount} has been debited from your account acc:{self.acct_no} and Your current balance is {self.balance}.')
        else :
            print(f'Insufficient Balance in account, acc : {self.acct_no}.')
    
    def transfer_bal(self,Amount,target_acct):
        if Amount> 0 :
            print(f'{Amount} has been transfered to {target_acct}.')
    
    
    def check_bal(self):
        print(f'Your balance is : {self.balance}')
```
